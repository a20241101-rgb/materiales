```python
#Luego de tener o imprimir el texto de la variable libro, convertimos todo el texto a minúscula con el código:
#creamos una variable llamado texto_min que guardará el texto en minúscula:
texto_min = libro.lower()
#imprimir el código para que, dentro de la variable, todo el texto sea minúscula.
print(texto_min)
```

```python
#Ahora eliminemos dentro del texto en minúscula todos los signos de puntuación:
#creamos una nueva variable que guardará el texto sin los signos de puntuación:
texto_depurado=texto_min.replace("-","").replace('"',"").replace(".", "").replace("—", "").replace("«", "").replace("»","").replace(",", "").replace("¿","").replace("?", "").replace("¡","").replace("!", "").replace("''","").replace(":", "").replace(";","").replace("(", "").replace(")","").replace("“","").replace("”", "")
#ejecutamos o imprimimos la variable de texto_depurado donde se guarda el texto con los signos ya eliminados:
print(texto_depurado)
```

```python
#CALCULAR LAS FRECUENCIAS DE LA LISTA_DEPURADA:
#Crear el diccionario de frecuencias llamado frecuencias_libro:
#creamos la variable que guarde el diccionario vacío:
frecuencias_libro=dict()
#usamos el bucle for para revisar cada palabra contenida en lista_libro:
for palabra in libro_depurado:
    if palabra in frecuencias_libro:#se evalua si la palabra que está revisando ya existe como clave o key en el diccionario
        frecuencias_libro[palabra]=frecuencias_libro[palabra]+1 #si la palabra ya se encuentra en el diccionario, incrementamos en 1 a su frecuencia o valor
    else:
        frecuencias_libro[palabra]=1 #si la palabra no existe en el diccionario, le agregamos como clave el valor de 1
print(frecuencias_libro) #Imprimimos en pantalla el diccionario final con las palabras (claves) y sus respectivas frecuencias (valores)
```

## **JUSTIFICACIÓN DEL PROCESAMIENTO**
Se convirtió todo el primer capítulo del libro de Mujercitas en minúsculas, debido a que se simplificaba el proceso para la eliminación de los stopwords al funcionar de manera efectiva sin dejar escapar conectores o palabras innecesarias que comenzaban en una oración. Asimismo, se eliminaron los números, signos de puntuación estructural, delimitadores de diálogo e incisos, signos de entonación y variados tipos de comillas y caracteres tipográficos. La eliminación de estos caracteres fue indispensable para impedir que se pegaran a las palabras adyacentes y se registraran las mismas palabras como distintas (como "mamá" con "mamá,"), en cambio se conservaron las tildes y las palabras para respetar la ortografía y el significado. Por otro lado, el criterio para seleccionar las stopwords se basó en excluir palabras de función estrictamente gramatical (artículos, preposiciones, conjunciones y pronombres) y algunos conectores que no aportan contenido semántico. En conjunto, estas decisiones eliminaron el ruido gramatical, permitiendo que la lista de frecuencias y la de palabras final resalten de forma clara los personajes y las acciones concretas que realizan. Estas decisiones influyeron de manera determinante en los resultados, puesto que al eliminar el contenido gramatical innecesario y consolidar las frecuencias reales, se evitó que conectores vacíos dominaran el conteo, logrando que la nube de palabras final resalte con la mayor escala visual a los personajes como Jo, Meg, Beth, Amy y la mamá, así como los temas o hechos verdaderamente representativos del capítulo elegido como navidad, regalos, dinero, trabajo y familia,  y a los temas verdaderamente representativos del capítulo como navidad, regalos, dinero y trabajo.
